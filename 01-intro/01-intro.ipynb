#Array Operations

```python
#Element-wise Operations
np.random.seed(3)
arr=np.random.rand(5)
print(arr)
arr=arr+1
arr=arr*2
arr
```

```python
#Element-wise Operations with Two Arrays of same size
arr1=np.zeros(5) #float [0. 0. 0. 0. 0.]
arr2=np.ones(5) #float [1. 1. 1. 1. 1.]
arr3=np.full(5,5) # integers [5 5 5 5 5]
arr1+=arr2
arr2*=arr3
# we get error if we choose arr3/=arr1 ->tries to put those floats back into the existing integer arr3.
arr3=(arr3/arr1).astype(int) #creates a new float array and assigns it to arr3, so it works.
print(arr3)
arr3=arr1/arr3
print(arr1)
print(arr2)
arr3




```

```python
#Comparison Operations
arr=np.array([1,2,3,4,5])
greater_than_2=arr>2
print(greater_than_2)
#Selecting Elements Based on Conditions
selected_elements=arr[arr>1]
selected_elements
#summary Operations
print(arr.min()) #min value of all array elements
print(arr.max()) #max value of all array elements
print(arr.sum()) #sum of all elemnets in array
print(arr.mean()) #avg of all array elements
print(arr.std()) #standard deviation of all array elements
```
